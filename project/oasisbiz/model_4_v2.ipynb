```python
import psycopg2
import pandas as pd
from sqlalchemy import create_engine,URL
```

```python
pd.set_option('display.max_columns',None)
```

```python
conn = psycopg2.connect(
  host='localhost',
  port=5432,
  dbname='postgres',
  user='postgres',
  password='postgres'
)
conn.set_session(autocommit=True)
cursor = conn.cursor()
```

```python
engine = create_engine(
  URL.create(
    drivername='postgresql+psycopg2',
    host='localhost',
    port=5432,
    database='postgres',
    username='postgres',
    password='postgres'
  )
)
```

```python
# read m1_total
cursor.execute(
  'select * from m4_total'
)
y_df = pd.DataFrame(
  cursor.fetchall(),
  columns=[col.name for col in cursor.description]
)
```

```python
y_df['deal_amount'] = y_df.deal_amount.astype('int64')
```

```python
y_df
```

```python
y_df.describe()
```

```python
y_df['by_arch_area_amount'] = y_df.deal_amount / y_df.arch_area
```

```python
deal_df_clean = y_df[0:0]

for sig in y_df['sig_cd'].unique():
  tmp_df = y_df[y_df['sig_cd'] == sig]
  # IQR 계산
  Q1 = tmp_df['by_arch_area_amount'].quantile(0.25)
  Q3 = tmp_df['by_arch_area_amount'].quantile(0.75)
  IQR = Q3 - Q1

  # # 이상치 제거
  deal_df_clean = pd.concat([
    deal_df_clean,
    tmp_df[~((tmp_df['by_arch_area_amount'] < (Q1 - 1.5 * IQR)) | (tmp_df['by_arch_area_amount'] > (Q3 + 1.5 * IQR)))]
  ])
```

```python
deal_df_clean['by_arch_area_amount'].describe()
```

```python
y_final_df = deal_df_clean[deal_df_clean['subway_dist'] <= 1000]
```

```python
y_final_df.to_sql(
  'm4_total_v2',
  engine,
  index=False,
  if_exists='replace'
)
```

```python
y_final_df.to_csv(
  'm4_data_06.29.csv',
  sep=',',
  index=False
)
```
